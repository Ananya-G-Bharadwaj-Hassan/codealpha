```python
#lineplot for analyzing unemployment rate over the years
plt.figure(figsize=(15,5))
sns.lineplot(x='Date', y='Estimated Unemployment Rate (%)', data=data1)
plt.title("Unemployment Rate Over Time")
plt.show()
```

The above line graph shows the unemployment rate over different time. Insights are as follows :

1. Before March, 2020 unemployment rate was very low (below 10%)
2. After March 2020, rate of unemployment gradually increased (upto 25%)
3. Tha major reason for this increase can be the effect of Covid-19

```python
#bar graph to know unemployment rate state wise
plt.figure(figsize=(14,7))
sns.barplot(x='Region', y='Estimated Unemployment Rate (%)', data=data1)
plt.xticks(rotation=90)
plt.show()
```

The above bar graph shows the unemployment rate state wise. Insights are:

1. States like assam, Meghalaya, Odisha etc have a low unemployment rate (below 10%)
2. This may be due to low population growth and very less opportunities
3. States like Tripura, Haryana, Jharkhand have a high unemployment rate (nearly 25% or more)
4. This may be due to high population growth and high competition

```python
#histogram to show distribution of data
sns.histplot(data1['Estimated Unemployment Rate (%)'], bins=20)
plt.show()
```

```python
#top 10 states with highest unemployement rate
state=data2.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)
state.head(10)
```

```python
#bar graph depicting the same
state.head(10).plot(kind='bar')
plt.show()
```

The above bar graph shows top 10 states with highest unemployment rate. Reasons being:

1. COVID-19 lockdowns caused a sharp rise in unemployment across most states.
2. States dependent on tourism, manufacturing, retail and services experienced larger increases in unemployment.
3. States with limited industrial development generally had higher average unemployment due to fewer employment opportunities.
4. Areas with high population growth faced greater pressure on local job markets.

```python
#boxplot to identify unemployment rate area wise that is rural and urban
sns.boxplot(x='Area', y='Estimated Unemployment Rate (%)', data=data1)
plt.show()
```

The boxplot shows that urban areas have a higher median unemployment rate than rural areas, along with greater variation in unemployment levels. 

This may be because urban economies depend heavily on industries, retail, tourism and service sectors, which were significantly affected during the COVID-19 lockdowns. 

Rural areas were relatively less affected as agriculture continued to provide employment and government schemes such as MGNREGA supported rural workers.

```python
#boxplot to identify unemployment rate region wise that is south, north, east and west
sns.boxplot(x='Region.1', y='Estimated Unemployment Rate (%)', data=data2)
plt.show()
```

The regional boxplot indicates that the North and East regions generally have higher unemployment rates, while the South and West show relatively lower median unemployment. 

These differences may be due to variations in industrial development, employment opportunities and economic activities across regions. 

The outliers suggest that some states were more severely affected by COVID-19 and lockdown measures.
